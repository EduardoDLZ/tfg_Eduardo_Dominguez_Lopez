```python
from pathlib import Path
import os

def find_root_dir(marker='tfg'):
    p = Path.cwd()
    for candidate in [p] + list(p.parents):
        if candidate.name == marker:
            return candidate.resolve()
        if (candidate / marker).is_dir():
            return (candidate / marker).resolve()
    raise FileNotFoundError(f"Could not find '{marker}' folder in {Path.cwd()} or its parents.")

os.chdir(find_root_dir('tfg'))
```

## Data Quality Assessment

```python
from src.preprocessing import clean_dataset, assess_data_quality, compare_datasets, save, load_dataset
from src.utils import intro
```

```python
def preprocessing(
    dataset_path: str | Path = None,
    invalid_negative_columns: list[str] = None,
    categorical_columns: list[str] = None,
):
    """
    Load, assess, clean, and compare a dataset
    """

    intro("Dataset Preprocessing")
    df = load_dataset(dataset_path)

    print(f"Dataset path: {Path(dataset_path).resolve()}")
    print(f"Original dataset shape: {df.shape}")

    quality_report = assess_data_quality(df)
    df_clean = clean_dataset(
        report=quality_report,
        df=df,
        invalid_negative_columns=invalid_negative_columns,
        categorical_columns=categorical_columns,
    )
    compare_datasets(df, df_clean)
    return df_clean
```

### 8k Baseline

```python
BASELINE_DATASET = "stage1/reference-8k-baseline.parquet"
REFERENCE_DATASET_PATH = Path("data/intermediate/reference-8k.parquet")
INVALID_NEGATIVE_COLUMNS = [
    "Flow IAT Min",
]

df_clean = preprocessing(REFERENCE_DATASET_PATH, INVALID_NEGATIVE_COLUMNS)
save(BASELINE_DATASET, df_clean)
```

### 225k Baseline

```python
FULL_DATASET = "stage2/full-225k-baseline.parquet"
FULL_DATASET_PATH = Path("data/intermediate/full-225k.parquet")
INVALID_NEGATIVE_COLUMNS = [
    "Flow Duration",
    "Flow Bytes/s",
    "Flow Packets/s",
    "Flow IAT Mean",
    "Flow IAT Max",
    "Flow IAT Min",
    "Fwd IAT Min",
]
df_clean = preprocessing(FULL_DATASET_PATH, INVALID_NEGATIVE_COLUMNS)
save(FULL_DATASET, df_clean)
```
